```python
# 定义sigmoid函数
def sigmoid(x):
    return 1 / (1 + np.exp(-x))#这是公式定义
```

```python
# 多层神经网络实现
x = np.array([[1,2]])
print(f'x为{x}')
w = np.array([[1,2,3],[5,6,7]])
print(f'w为{w}')
b = np.array([[1,1,1]])
print(f'b为{b}')
A1 = np.dot(x,w)+b
print(f'a1为{A1}')
A1.shape
# 隐藏层用激活函数
Z1 = sigmoid(A1)
print(Z1)
```

第二层实现


```python
W2 = np.array([[0.1, 0.4], [0.2, 0.5], [0.3, 0.6]])
B2 = np.array([0.1, 0.2])
print(Z1.shape) 
print(W2.shape) 
print(B2.shape) 
A2 = np.dot(Z1, W2) + B2
Z2 = sigmoid(A2)
```

```python
# 第二层到输出层实现，这里的激活函数可以变化一下
def identity_function(x):
    return x
W3 = np.array([[0.1, 0.3], [0.2, 0.4]])
B3 = np.array([0.1, 0.2])
A3 = np.dot(Z2, W3) + B3
Y = identity_function(A3) # 或者Y = A3
"""
一般地，回归问题可以使用恒等函数，
二元分类问题可以使用 sigmoid 函数，
多元分类问题可以使用 softmax 函数
"""
```

```python
# 代码小结
def init_network():#初始结构
    network = {}
    network['W1'] = np.array([[0.1, 0.3, 0.5], [0.2, 0.4, 0.6]])
    network['b1'] = np.array([0.1, 0.2, 0.3])
    network['W2'] = np.array([[0.1, 0.4], [0.2, 0.5], [0.3, 0.6]])
    network['b2'] = np.array([0.1, 0.2])
    network['W3'] = np.array([[0.1, 0.3], [0.2, 0.4]])
    network['b3'] = np.array([0.1, 0.2])
    return network
def forward(network, x):#前向传播由此而来
    W1, W2, W3 = network['W1'], network['W2'], network['W3']
    b1, b2, b3 = network['b1'], network['b2'], network['b3']
    a1 = np.dot(x, W1) + b1
    z1 = sigmoid(a1)
    a2 = np.dot(z1, W2) + b2
    z2 = sigmoid(a2)
    a3 = np.dot(z2, W3) + b3
    y = identity_function(a3)
    return y
network = init_network()
x = np.array([1.0, 0.5])
y = forward(network, x)
print(y) # [ 0.31682708  0.69627909]
```
