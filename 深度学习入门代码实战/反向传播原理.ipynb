````python
import numpy as np

# ==========================================
# 1. 初始化数据和参数
# ==========================================
# 最简单的单层感知机模型，只有一个输入节点和一个输出节点
# y=w*x
# 输入数据 (x) 和 真实标签 (y_true)
x = 2.0
y_true = 10.0

# 初始化权重 (w)，随便给一个初始值，比如 3.0
# 在实际神经网络中，这里通常是 np.random.randn()
w = 10.0

# 学习率 (learning rate)，控制每次参数更新的步伐大小
learning_rate = 0.1

print(f"--- 初始状态 ---")
print(f"输入 x: {x}, 目标 y: {y_true}")
print(f"初始权重 w: {w}")
print("-" * 30)

# ==========================================
# 2. 正向传播 (Forward Propagation)
# ==========================================

# 计算预测值 (y_pred)
# 公式: y_pred = x * w
y_pred = w * x

# 计算损失 (Loss)
# 我们使用平方误差公式: Loss = (y_pred - y_true)^2
loss = (y_pred - y_true) ** 2

print(f"--- 第一轮正向传播 ---")
print(f"预测值 (y_pred): {y_pred:.2f}")
print(f"损失值 (Loss): {loss:.2f}")

# ==========================================
# 3. 反向传播 (Backward Propagation)
# ==========================================
# 我们需要利用链式法则计算 Loss 对 w 的导数 (梯度)
# 链式法则公式: dLoss/dw = (dLoss/dy_pred) * (dy_pred/dw)

# 第一环: Loss 对 y_pred 的导数
# Loss = (y_pred - y_true)^2  => 导数是 2 * (y_pred - y_true)
grad_y_pred = 2 * (y_pred - y_true)

# 第二环: y_pred 对 w 的导数
# y_pred = x * w => 导数是 x
grad_w = x

# 最终梯度: Loss 对 w 的导数
gradient = grad_y_pred * grad_w

print(f"\n--- 第一轮反向传播 ---")
print(f"梯度 (Gradient): {gradient:.2f}") 
# 这里的 -16.0 和我们需要调整的方向一致（负数代表需要增加权重）

# ==========================================
# 4. 参数更新 (Optimizer Step)
# ==========================================

# 更新权重公式: w_new = w_old - (学习率 * 梯度)
w = w - (learning_rate * gradient)

print(f"\n--- 参数更新后 ---")
print(f"更新后的权重 w: {w:.2f}") 
# 结果应该是 4.60，完全符合刚才的手算

print("\n" + "="*30 + "\n")

# ==========================================
# 5. 自动化循环训练 (Training Loop)
# ==========================================
n =10
print(f'现在让模型自动训练 {n} 轮...')

# 重置权重以便演示
w = 3.0 

for epoch in range(1, n):
    # 1. 正向传播
    y_pred = w * x
    loss = (y_pred - y_true) ** 2
    
    # 2. 反向传播 (把上面的步骤合并写)
    # dL/dw = 2 * (y_pred - y_true) * x
    gradient = 2 * (y_pred - y_true) * x
    
    # 3. 更新参数
    w = w - (learning_rate * gradient)
    
    print(f"Epoch {epoch}: 预测值={y_pred:.4f}, Loss={loss:.4f}, 权重w更新为={w:.4f}")

print(f"\n最终结果: 当输入 x={x} 时，模型预测 {w*x:.4f} (非常接近 {y_true})")
````

````python
import numpy as np
# y = w*x + b
# ==========================
# 1. 初始化
# ==========================
x = np.random.randint(1,10)  # 随机输入
print(f"输入 x: {x}")
y_true = 100.0

w = 3.0
b = 1.0  
lr = 0.01  
epochs = 100 # 训练轮数

print(f"--- 开始训练 ---")
print(f"初始状态: w={w:.4f}, b={b:.4f}, 目标 y={y_true}")
print(f"初始预测: {w*x+b:.4f} (距离目标还差 {100 - (w*x+b):.4f})")
print("-" * 50)

# ==========================
# 2. 循环训练
# ==========================
for epoch in range(epochs):
    
    # --- [Step A: 正向传播] ---
    a = w * x
    y_pred = a + b 
    loss = (y_pred - y_true) ** 2
    
    # --- [Step B: 反向传播] ---
    # 1. Loss 对 y_pred 的导数
    grad_y_pred = 2 * (y_pred - y_true)
    
    # 2. 加法节点分流 (y_pred = a + b)  加法是“公平分发器”，不管谁进来，梯度都乘以 1
    grad_b = grad_y_pred * 1
    grad_a = grad_y_pred * 1
    
    # 3. 乘法节点反向 (a = w * x)
    grad_w = grad_a * x
    
    # --- [Step C: 参数更新] ---
    w = w - lr * grad_w
    b = b - lr * grad_b
    
    # --- [打印进度] ---
    # 每 10 轮打印一次，避免刷屏
    if (epoch + 1) % 10 == 0:
        print(f"Epoch {epoch+1:03d} | Loss: {loss:.4f} | 预测值: {y_pred:.4f} | w: {w:.4f} | b: {b:.4f}")

# ==========================
# 3. 最终结果验证
# ==========================
print("-" * 50)
print(f"--- 训练结束 ---")
print(f"最终参数: w={w:.4f}, b={b:.4f}")

# 验证
final_pred = w * x + b
print(f"最终验证: {w:.4f} * {x} + {b:.4f} = {final_pred:.4f}")
print(f"目标值是: {y_true}")

# 思考题：
# 你会发现 w 变成了约 24.5，b 变成了约 26.3
# 24.5 * 3 + 26.3 ≈ 100
# 其实 w=33, b=1 也可以；w=0, b=100 也可以。
# 为什么模型选了这一组？这取决于初始值和梯度的比例 (x=3, 所以 w 的梯度永远是 b 的 3 倍，w 跑得比 b 快)。
````

````python
import numpy as np

# ==========================================
# 1. 准备数据 (Batch Data)
# ==========================================
# 我们用 numpy 数组来一次性存储所有点
# y=2*x
X = np.array([1.0, 2.0, 3.0])  # 输入
Y = np.array([2.0, 4.0, 6.0])  # 真实标签

# ==========================================
# 2. 初始化参数
# ==========================================
# 我们故意给一个很烂的初始值
w = 0.5  # 理想是 2.0
b = 1.0  # 理想是 0.0

lr = 0.05  # 学习率
epochs = 50 # 训练轮数

print(f"--- 训练开始 ---")
print(f"初始参数: w={w}, b={b}")
print("-" * 40)

# ==========================================
# 3. 循环训练 (Training Loop)
# ==========================================
for epoch in range(epochs):
    
    # --- [正向传播] ---
    # numpy 的广播机制允许我们直接对数组进行运算
    # 这里的 y_pred 也是一个包含3个数字的数组
    y_pred = w * X + b  
    
    # 计算损失 (MSE - Mean Squared Error)
    # 1. 算出每个点的误差平方
    errors = (y_pred - Y) ** 2
    # 2. 求平均
    loss = np.mean(errors)
    
    # --- [反向传播] ---
    # 我们需要求 Loss 对 w 和 b 的平均梯度
    # MSE公式展开是: L = 1/N * sum((wx+b - y)^2)
    
    # 1. 计算误差向量 (y_pred - y)
    diff = (y_pred - Y)
    
    # 2. 计算 w 的梯度
    # dL/dw = 2/N * sum((y_pred - y) * x)
    # 在 numpy 中，(diff * X) 就是对应元素相乘，然后 .mean() 包含了除以 N
    w_grad = 2 * np.mean(diff * X)
    
    # 3. 计算 b 的梯度
    # dL/db = 2/N * sum(y_pred - y)
    # 因为对 b 求导系数是 1，所以直接求误差的平均即可
    b_grad = 2 * np.mean(diff)
    
    # --- [参数更新] ---
    w = w - lr * w_grad
    b = b - lr * b_grad
    
    # 每隔 10 轮打印一次进度，观察变化
    if (epoch + 1) % 10 == 0 or epoch == 0:
        print(f"Epoch {epoch+1:02d} | Loss: {loss:.4f} | w: {w:.4f} | b: {b:.4f}")

print("-" * 40)
print("--- 训练结束 ---")
print(f"最终预测参数: w={w:.4f}, b={b:.4f}")
print(f"理想目标参数: w=2.0000, b=0.0000")

# 验证一下
test_x = 4
print(f"测试: 输入 {test_x}, 模型预测 {w*test_x + b:.2f} (应该是 8)")
````

### 第二部分：详细拆解 (结合例子)

这里你要主动拿出纸笔，或者用手比划，带入我们刚才的例子。

#### 1. 正向传播 (Forward Propagation)

- **面试官问**：正向传播在做什么？

- **你的回答**：

  > “正向传播是数据从输入层流向输出层的过程。它的核心目的是计算当前的预测值（Prediction）和损失（Loss）。
  >
  > 举个最简单的例子，假设模型是 $y = w \cdot x$。
  >
  > 如果输入 $x=2$，当前权重 $w=3$，正向传播就是计算 $2 \times 3 = 6$。
  >
  > 然后拿着这个 6 去和真实标签（比如 10）做对比，算出误差（Loss），比如均方误差是 16。
  >
  > 在这个阶段，参数 $w$ 是不更新的，我们只是在评估当前模型表现得怎么样。”

#### 2. 反向传播 (Backward Propagation)

- **面试官问**：那反向传播呢？为什么它是核心？

- **你的回答**：

  > “反向传播是训练的灵魂。它的目的是计算损失函数相对于每个参数的梯度，告诉参数该怎么变才能降低误差。
  >
  > 接上面的例子，既然误差是 16，反向传播就会从 Loss 节点出发，利用**链式法则（Chain Rule）**由后向前推导。
  >
  > 它的逻辑是：‘想要 Loss 变小，预测值 $\hat{y}$ 得变大；想要 $\hat{y}$ 变大，权重 $w$ 就得变大’。
  >
  > 数学上，这相当于计算 $\frac{\partial L}{\partial w}$。在刚才的例子里，我们算出梯度是 -16。这个梯度就是反向传播传递回来的‘信号’，指导优化器去更新权重。”

### 第三部分：高分亮点 (Bonus Points)

如果想让面试官眼前一亮，你要提到**“计算图”**和**“效率”**。

**话术建议：**

> “其实在 PyTorch 或 TensorFlow 等框架的底层，这两个过程是在一张**计算图（Computation Graph）**上完成的。
>
> - **正向传播**构建了图的节点和数据流。
> - **反向传播**的高明之处在于它是一种**高效的动态规划**。它避免了重复计算，每个节点只需要处理上游传来的梯度（Gradient）乘以本地的导数（Local Gradient），就像接力赛传棒一样。
>
> 比如在 $y=wx+b$ 中，加法节点就是一个‘梯度分发器’，把梯度原封不动传给 $w$ 和 $b$ 所在的支路；而乘法节点则会把梯度进行缩放。这种机制让我们可以训练几百层深的网络而不用手算复杂的导数。”

### 总结一张表 (Cheat Sheet)

面试前扫一眼这个表格，心里就有底了：

| **维度**     | **正向传播 (Forward)**                            | **反向传播 (Backward)**              |
| ------------ | ------------------------------------------------- | ------------------------------------ |
| **方向**     | 输入层 $\rightarrow$ 输出层                       | 损失节点 $\rightarrow$ 输入/权重层   |
| **核心动作** | 乘法、加法、激活函数                              | 求导、链式法则 (Chain Rule)          |
| **产出物**   | 预测值 ($\hat{y}$)、损失值 (Loss)                 | 梯度 (Gradients, $\nabla w$)         |
| **物理意义** | **做题**：给出答案并打分                          | **改错**：根据错题找原因             |
| **代码对应** | `y_pred = model(x)` `loss = criterion(y_pred, y)` | `loss.backward()` `optimizer.step()` |
| **关键点**   | 不涉及参数修改                                    | 不涉及预测值计算，只算方向           |

### 面试中的“必杀技”例子

如果面试官让你举例说明 **“链式法则”** 是怎么工作的，直接甩出这个例子：

> “好，比如 $Loss = (wx - y)^2$。我们把它看作三层：
>
> 1. 内层 $a = wx$
> 2. 中层 error $e = a - y$
> 3. 外层 $L = e^2$
>
> 反向传播就是剥洋葱：
>
> 先算外层导数 $\frac{dL}{de}$，再乘以中层导数 $\frac{de}{da}$，最后乘以内层导数 $\frac{da}{dw}$。
>
> 就像我刚才代码里演示的，最终梯度就是把这三部分乘起来，这就叫链式替换。”

这一套下来，既有理论深度，又有具体例子，面试官通常会对你的基础知识非常满意。
